```python
df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/Volumes/databricks-pyspark/databricks-pyspark-schema/internal-v1/emp_1.csv")
```

```python
df.show()
```

```python
df2 = df.filter(df['address'] == 'india').show()
```

```python
df.filter(df['address'] != 'india').show()
```

```python
df.filter((df['name'] == 'manish') | (df['salary'] == 10000)).show()
```

### name starts with R Filter condition

```python
df.filter(df['name'].startswith('r')).show()
```

### name end with i using filter function

```python
df.filter(df['name'].endswith('l')).show()
```

### names contains ul filter condition

```python
df.filter(df['name'].like('%ul%')).show()
```
