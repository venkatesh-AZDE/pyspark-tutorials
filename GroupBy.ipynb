```python
df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/Volumes/databricks-pyspark/databricks-pyspark-schema/internal-v1/emp_1.csv")
```

```python
df.show()
```

```python
from pyspark.sql.functions import first,sum
```

agg is aggregate function used to apply one or more functions

```python
df.groupBy("address","emp_id") \
    .agg( 
         first("name").alias("firstname"),
         sum("salary").alias("agg_salary")
    ) \
    .show()
```

### max salary

```python
df.groupBy("address").max("salary").show()
```

```python
from pyspark.sql.functions import max
```

```python
df.groupBy("address") \
    .agg(
        max("salary").alias("max_salary")

    ).show()
```
