```python
df_emp = spark.read.format("csv").option("header",True).option("inferSchema",True).load("/Volumes/databricks-pyspark/databricks-pyspark-schema/internal-v1/employee (1).csv")
```

```python
new_record = spark.createDataFrame([(10,"John",5000,"NY")],["emp_id","name","salary","loc"])
df_empnew = df_emp.union(new_record)
```

```python
df_empnew.show()
```

```python
df_dep = spark.read.format("csv").option("inferSchema",True) \
    .option("header",True) \
        .load("/Volumes/databricks-pyspark/databricks-pyspark-schema/internal-v1/department.csv")
```

```python
df_dep.show()
```

```python
df_join = df_empnew.join(df_dep,df_empnew.emp_id == df_dep.user,"inner")
```
