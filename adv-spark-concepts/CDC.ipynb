```python
data = [
    (1,"John","Chicago","updated","2026-05-02 10:11:12"),
    (1,"John","lasvegas","updated","2026-05-02 10:12:12"),
    (1,"John","wqashington","updated","2026-05-02 10:13:12"),

]
cdc_df = spark.createDataFrame(data, ["emp_id","name","country","dep","current_time"])
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number,col

window_spec = Window.partitionBy("emp_id") \
                    .orderBy(col("current_time").desc())

latest_cdc_df = cdc_df.withColumn(
    "rn",
    row_number().over(window_spec)
).filter("rn = 1") \
 .drop("rn")

latest_cdc_df.show()
```

```python
latest_cdc_df.write \
  .format("delta") \
  .mode("overwrite") \
  .saveAsTable("employee_cdc")
```
