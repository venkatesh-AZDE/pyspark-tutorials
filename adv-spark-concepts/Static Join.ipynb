```python
customer_data = [
    (1, "John", "USA"),
    (2, "Mary", "UK"),
    (3, "Mike", "Canada")
]

customer_df = spark.createDataFrame(
    customer_data,
    ["customer_id","customer_name","city"]
)
```

```python
stream_df = spark.readStream.format("rate") \
                            .option("rowspersecond", 10) \
                            .option("numPartitions", 2) \
                            .load()
```

```python
from pyspark.sql.functions import *
```

```python
query = stream_df.select(
    (col("value") % 3 + 1).alias("customer_id"),
    col("value").alias("order_id")
)
```

```python
joined_df = query.join(
    customer_df,
    customer_df.id == query.customer_id,'left'
)
```

```python
joined_df.show()
```

```python
query = joined_df.writeStream \
    .format("memory") \
    .queryName("joined_output") \
    .outputMode("append") \
    .trigger(availableNow=True)
```

```python
customer_data = [
    (1, "John", "USA"),
    (2, "Mary", "UK"),
    (3, "Mike", "Canada")
]

customer_df = spark.createDataFrame(
    customer_data,
    ["customer_id","customer_name","city"]
)
```

```python
stream_df = spark.readStream \
    .format("rate") \
    .option("rowsPerSecond", 1) \
    .load()
```

```python
from pyspark.sql.functions import *

orders_df = stream_df.select(
    (col("value") % 3 + 1).alias("customer_id"),
    col("value").alias("order_id")
)
```

```python
from pyspark.sql.functions import *

orders_df = stream_df.select(
    (col("value") % 3 + 1).alias("customer_id"),
    col("value").alias("order_id")
)
```

```python
joined_df = orders_df.join(
    customer_df,
    on="customer_id",
    how="inner"
)
```

```python
query = joined_df.writeStream \
    .format("memory") \
    .queryName("joined_output") \
    .outputMode("append") \
    .trigger(availableNow=True)
```
