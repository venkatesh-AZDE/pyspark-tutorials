```python
data = [
    (1,"manish","india",25000),
    (2,"john","USA",30000),
    (3,"sara","USA",20000),
    (4,"mike","india",10000),
    (5,"raj","india",15000),
    (6,"sara","USA",20000)
]
schema = ["id","name","country","salary"]
df = spark.createDataFrame(data,schema)
```

```python
india_df = df.filter(df.country == "india")
```

```python
india_df.count()
```

```python
india_df.show()
```
