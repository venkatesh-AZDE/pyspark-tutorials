```python
base_path = "/Volumes/databricks-pyspark/spark_ns/ns_volume/autoloader_demo"
```

```python
source_path = base_path + "/source_path"
schema_path = base_path + "/schema_path"
checkpoint_path = base_path + "/checkpoint_path"
```

```python
dbutils.fs.put(
    source_path + "/employee_01.csv",
    """emp_id,name,salary
1,John,10000
2,Mary,20000
3,Mike,30000
""",
    True
)
```

```python
input_df = spark.readStream \
                .format("cloudFiles") \
                    .option("cloudFiles.format","csv") \
                        .option("cloudFiles.schemaLocation",schema_path) \
                            .option("header","True") \
                                .option("inferSchema","True") \
                                    .load(source_path)
```

```python
query = input_df.writeStream \
                .format("delta") \
                .option("checkpointLocation",checkpoint_path) \
                    .trigger(availableNow=True) \
                        .toTable("employee_bronzev1")
```

```python
spark.table("employee_bronzev1")
```
