```python
data = [
    (1, "manish", 20000, "india"),
    (2, "john", 30000, "usa"),
    (3, "sara", 25000, "uk"),
    (4, "ravi", 40000, "india"),
    (5, "amit", 35000, "india"),
    (6, "david", 45000, "usa"),
    (7, "priya", 50000, "india"),
    (8, "tom", 28000, "uk")
]

columns = ["emp_id", "name", "salary", "loc"]

df = spark.createDataFrame(data, columns)

df.show()
```

```python
from pyspark.sql.functions import spark_partition_id,count
df.withColumn("partition_id",spark_partition_id()) \
                         .groupBy("partition_id") \
                             .agg(count('*').alias("rows_count")).show()
```

```python
df = df.repartition(4)
df = df.withColumn("partition_id",spark_partition_id()) \
                         .groupBy("partition_id") \
                             .agg(count('*').alias("rows_count"))
```

```python
data = [
    (1, "manish", 20000, "india"),
    (2, "john", 30000, "usa"),
    (3, "sara", 25000, "uk"),
    (4, "ravi", 40000, "india"),
    (5, "amit", 35000, "india"),
    (6, "david", 45000, "usa"),
    (7, "priya", 50000, "india"),
    (8, "tom", 28000, "uk")
]

columns = ["emp_id", "name", "salary", "loc"]

df = spark.createDataFrame(data, columns)
```

```python
#check how many partion does df has 
df.withColumn("partition_id",spark_partition_id()) \
             .groupBy("partition_id") \
             .agg(count('*')).alias("rows_count").show()

             
```

```python
#no partitions detects that means all data reside on single partition 
#now we want to create 4 partitions
```

```python
re_df = df.repartition(4)
#check how many partition does re_df has
re_df.withColumn('partition_id',spark_partition_id()) \
    .groupBy('partition_id') \
        .agg(count('*').alias("rows_count")).show()
#now we have 4 partitions
```

```python
co_df = re_df.coalesce(2)
co_df.withColumn('partition_id',spark_partition_id()) \
    .groupBy('partition_id') \
        .agg(count('*').alias("rows_count")).show()
#now we have 2 partitions
```

```python
co_df.write.mode("overwrite").format("parquet").save("/Volumes/databricks-pyspark/databricks-pyspark-schema/internal-v1/output/colesce_output/")
```
