```python
from pyspark.sql.functions import col
from pyspark.sql.types import TimestampType

# Create DataFrame with fromdate and todate columns (timestamp)
df_dates = spark.createDataFrame([
    ('2025-07-01 07:01:12', '2025-07-01 07:01:12')
], ['fromdate', 'todate'])

# Cast columns to timestamp
df_dates = df_dates.withColumn('fromdate', col('fromdate').cast(TimestampType())) \
                   .withColumn('todate', col('todate').cast(TimestampType()))

# Add 7 days to todate, keeping timestamp
df_dates = df_dates.withColumn('todate_plus_7', expr("date_add(DAY, 7, todate)"))

# Calculate date difference
df_diff = df_dates.withColumn('date_difference', expr("todate_plus_7 - fromdate"))

display(df_dates)
```
