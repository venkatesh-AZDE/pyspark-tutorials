```python
data = [
    (1, 'John', 50000, 'New York'),
    (2, 'Jane', 60000, 'Los Angeles'),
    (3, 'Bob', 55000, 'Chicago')
]

df = spark.createDataFrame(data, ['id','name','salary','loc'])
```

```python
df.show()
```

```python
from pyspark.sql.functions import upper  
```

```python
def uppername(input_df):
    return input_df.withColumn('name', upper(input_df["name"])) 
    
```

```python
display(df.transform(uppername))
```
