#### fillna or fill is used to replace null/None with zero or any literal

```python
df = spark.read.format("csv").option("header",True).option("inferSchema",True).load("/Volumes/databricks-pyspark/databricks-pyspark-schema/internal-v1/sample.csv")
```

```python
df.show()
```

```python
df.na.fill("",["city"]).na.fill(0,["population"]).show()
```
