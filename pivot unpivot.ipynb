```python
data = [
    (1,"orange","","usa"),(2,"banana","","canada"),(3,"banana",2000,"india"),(4,"orange",1000,"india"),(5,"mango",3000,"india"),(6,"mango",2000,"usa"),(7,"apple",1000,"india"),(8,"apple",3000,"canada")
]
schema = ["id","item","total","country"]
df = spark.createDataFrame(data, schema)
df.show()
```

```python
from pyspark.sql.functions import col
```

```python
df_pivot = df.withColumn("total", col("total").try_cast('double')).groupBy("item").pivot("country").sum("total").show()
```

```python
from pyspark.sql.functions import expr
```

```python
df2 = df_pivot.select(expr("item as dept"), expr("canada as emp_name"), expr("usa as loc"), expr("india as salary"))
```
