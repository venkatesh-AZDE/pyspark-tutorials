```python
from pyspark.sql.types import StructType,StructField
from pyspark.sql.types import IntegerType,StringType
from pyspark.sql.types import ArrayType
from pyspark.sql.functions import explode
```

```python
prepareSchema = StructType([
    StructField("userId",StringType(),True),
    StructField("emailAddress",StringType(),True),
    StructField("employeeCode",StringType(),True),
    StructField("firstName",StringType(),True),
    StructField("jobTitleName",StringType(),True),
    StructField("lastName",StringType(),True),
    StructField("phoneNumber",StringType(),True),
    StructField("preferredFullName",StringType(),True),
    StructField("region",StringType(),True),
])
```

```python
finalStruct = StructType([
    StructField("Employees",ArrayType(prepareSchema),True)
]
)
```

```python
df_json = spark.read.schema(finalStruct).option("multiLine",True).json("/Volumes/databricks-pyspark/databricks-pyspark-schema/internal-v1/employees.json")
```
