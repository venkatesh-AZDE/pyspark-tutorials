```python
data = [
    (1, 'John', 50000, 'New York'),
    (2, 'Jane', 60000, 'Los Angeles'),
    (3, 'Bob', 55000, 'Chicago'),
    (4, 'Alice', 70000, 'Houston'),
    (5, 'Mike', 45000, 'Miami'),
    (6, 'Sara', 65000, 'Dallas')
]

df = spark.createDataFrame(data, ['id', 'name', 'salary', 'loc'])
```

```python
df.createOrReplaceTempView("employeesTempView")
```

```python
%sql
select * from employeesTempView;
```

```python
%sql
select max(salary) from employeesTempView;
```
