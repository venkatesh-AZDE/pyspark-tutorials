```python
import pyspark
```

```python
data1 = [(1,'sales','rohn','NJ',100000,34),(2,'hr','john','NY',120000,36),(3,'finance','jack','CA',90000,32),(4,'IT','jill','TX',80000,25),(5,'IT','jill','TX',80000,25),(8,'IT','jill','TX',80000,25)]
df1 = spark.createDataFrame(data1,['id','dept','emp_name','loca','salary','age'])
```

```python
df1.printSchema()
```

```python
data2 = [(7,'IT','jill','TX',80000,25),(8,'IT','jill','TX',80000,25)]
df2 = spark.createDataFrame(data=data2,schema=["id","dept","emp_name","loca","salary","age"])
```

```python
df2.printSchema()
```

```python
df_union = df1.union(df2)
```

```python
df_union.show()
```

```python
df_unionall = df1.unionAll(df2)
```

```python
df_unionall.show()
```

##### union removes all duplicate records 
##### unionall keeps all records
