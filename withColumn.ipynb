```python
df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/Volumes/databricks-pyspark/databricks-pyspark-schema/internal-v1/emp_1.csv")
```

```python
df = df.withColumn("salary1",df.salary*10)
```

```python
df.show()
```

### new column using a constant


```python
from pyspark.sql.functions import lit,col
```

```python
df2 = df.withColumn("country",lit("India"))
```

```python
df2.show()
```

### rename column data type

```python
df3 = df.withColumn("salary",df.salary.cast("Integer"))
```

```python
df3.show()
```

```python
df3.printSchema()
```

### withcolumn renamed

```python
df4 = df3.withColumnRenamed("loc","location")
df4.show()
```
